# === NumPy Retail Analytics Assignment (50 - 100) ===

```python
import numpy as np

# Store names
stores = np.array([
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Mumbai",
    "Pune"
])

# Product categories
categories = np.array([
    "Electronics",
    "Groceries",
    "Clothing",
    "Furniture",
    "Beauty",
    "Sports"
])

# Sales revenue in rupees
# Rows represent stores
# Columns represent product categories

sales_data = np.array([
    [850000, 420000, 310000, 275000, 190000, 225000],
    [780000, 460000, 295000, 310000, 210000, 240000],
    [920000, 390000, 340000, 260000, 185000, 215000],
    [1050000, 510000, 375000, 345000, 250000, 290000],
    [690000, 405000, 280000, 225000, 175000, 205000]
])

# Cost incurred in rupees
# Rows represent stores
# Columns represent product categories

cost_data = np.array([
    [640000, 315000, 220000, 205000, 125000, 160000],
    [590000, 350000, 210000, 225000, 140000, 170000],
    [700000, 295000, 245000, 195000, 120000, 155000],
    [790000, 390000, 265000, 250000, 165000, 205000],
    [530000, 310000, 205000, 170000, 115000, 145000]
])

print("Stores:")
print(stores)

print("\nProduct Categories:")
print(categories)

print("\nSales Data:")
print(sales_data)

print("\nCost Data:")
print(cost_data)
```

# --- Question 51: Best Target-Performing Store ---

```python
# Create sales targets
sales_targets = np.array([
    800000, 450000, 300000,
    280000, 200000, 220000
])

# Create target achievement matrix
target_achieved = sales_data >= sales_targets

# Count achieved targets for each store
categories_achieved = np.sum(target_achieved, axis=1)
```

```python
# Find the store that achieved the highest number of category targets.

# Find the store with the highest number of targets achieved
highest_target_index = np.argmax(categories_achieved)

highest_target_store = stores[highest_target_index]
highest_target_count = categories_achieved[highest_target_index]

print("Store:", highest_target_store)
print("Number of Targets Achieved:", highest_target_count)
```

## ---- Question 52: Target Achievement by Category ---

```python
#Calculate how many stores achieved the target for each product category.

# Calculate how many stores achieved the target for each category
stores_achieved = np.sum(target_achieved, axis=0)

# Display category names with number of stores that achieved the target
for category, count in zip(categories, stores_achieved):
    print(category + ":", count)
```

# --- Question 53: Sales Target Variance ---

```python
# Calculate the difference between actual sales and target sales for every store.

# Use:

# Variance = Actual Sales - Target Sales

# Calculate variance between actual sales and target sales
sales_variance = sales_data - sales_targets

print("Sales Variance:")
print(sales_variance)
```

# --- Question 54: Percentage Above or Below Target ---

```python
# Calculate the percentage variance from the target using:

# Percentage Variance =
# (Actual Sales - Target Sales) / Target Sales × 100

# Calculate percentage variance
percentage_variance = (
    (sales_data - sales_targets) / sales_targets
) * 100

# Round to 2 decimal places
percentage_variance = np.round(percentage_variance, 2)

print("Percentage Variance:")
print(percentage_variance)
```

# --- Question 55: Stores Missing Electronics Target ---

```python
# Identify stores that failed to achieve the Electronics sales target.

# Find stores that failed to achieve the Electronics target
failed_electronics = stores[sales_data[:, 0] < sales_targets[0]]

print("Stores that failed to achieve Electronics target:")
print(failed_electronics)
```

# --- Question 56: Stores Achieving All Targets ---

```python
# Check whether any store achieved its target in all six product categories.

# Use np.all().

# Check whether each store achieved all 6 category targets
all_targets_achieved = np.all(target_achieved, axis=1)

print("All Targets Achieved:")
print(all_targets_achieved)

# Display stores that achieved all targets
print("Stores that achieved all targets:")
print(stores[all_targets_achieved])
```

# --- Question 57: Stores Achieving At Least One Target ---

```python
#Identify stores that achieved at least one category target.

# Use np.any().

# Check whether each store achieved at least one target
at_least_one_target = np.any(target_achieved, axis=1)

print("At Least One Target Achieved:")
print(at_least_one_target)

# Display the stores
print("Stores that achieved at least one target:")
print(stores[at_least_one_target])
```

# --- Question 85: Outlier Detection ---

```python
# Treat any sales value greater than:

# Mean + 1.5 × Standard Deviation
# as a high-value outlier.

# Identify all such sales values.

# Calculate mean and standard deviation
mean_sales = np.mean(sales_data)
std_sales = np.std(sales_data)

# Calculate high-value outlier threshold
outlier_threshold = mean_sales + (1.5 * std_sales)

print("Outlier Threshold:", outlier_threshold)

# Find indexes of high-value outliers
store_indexes, category_indexes = np.where(
    sales_data > outlier_threshold
)

# Display the outliers
for store_index, category_index in zip(store_indexes, category_indexes):
    print(
        "Sales:", sales_data[store_index, category_index],
        "| Store:", stores[store_index],
        "| Category:", categories[category_index]
    )
```
