# --- Question 45: Premium Performance ---

```python
# A store-category combination is considered premium when:

# Sales are greater than ₹5,00,000
# Profit is greater than ₹1,50,000
# Find all premium-performing combinations.

# Find premium-performing combinations
premium_mask = np.logical_and(
    sales_data > 500000,
    profit_data > 150000
)

# Get store and category indexes
store_indexes, category_indexes = np.where(premium_mask)

# Display results
for store_index, category_index in zip(store_indexes, category_indexes):
    print(
        stores[store_index],
        "-",
        categories[category_index],
        "| Sales:", sales_data[store_index, category_index],
        "| Profit:", profit_data[store_index, category_index]
    )
```
