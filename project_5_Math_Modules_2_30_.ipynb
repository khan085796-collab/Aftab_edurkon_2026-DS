```python
## 4. Reverse a three-digit number using arithmetic operators.
num = [123]

n = num[0]

digit1 = n % 10
digit2 = (n // 10) % 10
digit3 = n // 100

reverse = digit1 * 100 + digit2 * 10 + digit3

print("Original number:", n)
print("Reversed number:", reverse)
```

```python
## 5. Find the last digit and first digit of a given number.
num = 123456789

# Find the last digit
last_digit = num % 10

# Find the first digit
first_digit = num
while first_digit >= 10:
    first_digit = first_digit // 10

print("First digit:", first_digit)
print("Last digit:", last_digit)
```

```python
## 21.Count how many bits are set to 1 in a number using bitwise operations.
# Input a number
num = int(input("Enter a number: "))

count = 0
temp = num

while temp > 0:
    if temp & 1:
        count += 1
    temp = temp >> 1

print("Number of set bits:", count)
```

```python
## 28.Build a simple calculator using arithmetic and comparison operators.

               #simple calculator..........


num1 = float(input("enter a number"))
num2 = float(input("enter a number"))
operator = input("Enter operator (+, -, *, /): ")

if operator == "+":
    print("Result =", num1 + num2)

elif operator == "-":
    print("Result =", num1 - num2)

elif operator == "*":
    print("Result =", num1 * num2)

elif operator == "/":
    if num2 == 0:
        print("Division by zero is not allowed.")
    else:
        print("Result =", num1 / num2)

else:
    print("Invalid operator!")
```

```python
## 30.Build a menu-driven calculator that uses arithmetic, comparison, logical, assignment, membership, identity, and bitwise operators wherever applicable.
# Menu-Driven Calculator

total_operations = 0

while True:
    print("\n===== MENU =====")
    print("1. Arithmetic Operations")
    print("2. Comparison Operations")
    print("3. Logical Operations")
    print("4. Membership Operations")
    print("5. Identity Operations")
    print("6. Bitwise Operations")
    print("7. Exit")

    choice = int(input("Enter your choice: "))

    total_operations += 1      # Assignment operator

    if choice == 1:
        a = int(input("Enter first number: "))
        b = int(input("Enter second number: "))

        print("Addition:", a + b)
        print("Subtraction:", a - b)
        print("Multiplication:", a * b)

        if b != 0:
            print("Division:", a / b)
            print("Floor Division:", a // b)
            print("Modulus:", a % b)
        else:
            print("Division by zero not allowed.")

        print("Power:", a ** b)

    elif choice == 2:
        a = int(input("Enter first number: "))
        b = int(input("Enter second number: "))

        print("a == b :", a == b)
        print("a != b :", a != b)
        print("a > b  :", a > b)
        print("a < b  :", a < b)
        print("a >= b :", a >= b)
        print("a <= b :", a <= b)

    elif choice == 3:
        x = int(input("Enter first number: "))
        y = int(input("Enter second number: "))

        print("x > 0 and y > 0 :", x > 0 and y > 0)
        print("x > 0 or y > 0  :", x > 0 or y > 0)

    elif choice == 4:
        text = input("Enter a string: ")
        ch = input("Enter a character: ")

        print(ch, "in string:", ch in text)
        print(ch, "not in string:", ch not in text)

    elif choice == 5:
        list1 = [1, 2, 3]
        list2 = [1, 2, 3]
        list3 = list1

        print("list1 is list2 :", list1 is list2)
        print("list1 is list3 :", list1 is list3)
        print("list1 is not list2 :", list1 is not list2)

    elif choice == 6:
        a = int(input("Enter first number: "))
        b = int(input("Enter second number: "))

        print("AND (&):", a & b)
        print("OR (|):", a | b)
        print("XOR (^):", a ^ b)
        print("Left Shift (a << 1):", a << 1)
        print("Right Shift (a >> 1):", a >> 1)

    elif choice == 7:
        print("Program Ended.")
        break

    else:
        print("Invalid Choice!")

print("Total menu selections:", total_operations)
```
