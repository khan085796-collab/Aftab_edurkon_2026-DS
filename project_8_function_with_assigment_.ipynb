```python
#Assignment 3: Calculator Using Functions
   #Create four separate functions:
#add()
#subtract()
#multiply()
#divide()
#Requirements
#Accept two numbers from the user.
#Ask the user to select an operation.
#Call the appropriate function.
#Handle division by zero.
#Print the calculated result.
#Example Input
#Enter first number: 12
#Enter second number: 4
#Enter operation: divide
#Expected Output
#Result: 3.0
#Hint
#Use if, elif, and else to select the function.

# Function to add two numbers
def add(num1, num2):
    return num1 + num2

# Function to subtract two numbers
def subtract(num1, num2):
    return num1 - num2

# Function to multiply two numbers
def multiply(num1, num2):
    return num1 * num2

# Function to divide two numbers
def divide(num1, num2):
    # Check for division by zero
    if num2 == 0:
        return "Error! Division by zero is not allowed."
    return num1 / num2

# Read the first number from the user
num1 = float(input("Enter first number: "))

# Read the second number from the user
num2 = float(input("Enter second number: "))

# Ask the user to choose an operation
operation = input("Enter operation (add, subtract, multiply, divide): ").lower()

# Call the appropriate function based on the user's choice
if operation == "add":
    result = add(num1, num2)
elif operation == "subtract":
    result = subtract(num1, num2)
elif operation == "multiply":
    result = multiply(num1, num2)
elif operation == "divide":
    result = divide(num1, num2)
else:
    result = "Invalid operation!"

# Display the result
print("Result:", result)

```

```python
#Assignment 5: Find the Largest Number

 #Write a function named find_largest() that accepts three numbers.

     #Requirements

      #Do not use the built-in max() function.
      #Compare the numbers using conditional statements.
      #Return the largest number.
      #Accept three values from the user.

# function to find the largest of three number:


def find_largest(num1, num2, num3):
    if num1 >= num2 and num1 >= num3:
        return num1
    elif num2 >= num1 and num2 >= num3:
        return num2
    else:
        return num3

# Accept three numbers from the user
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
num3 = float(input("Enter the third number: "))

# Call the function
largest = find_largest(num1, num2, num3)

# Print the result
print("Largest number is:", largest)
```

```python

##Assignment 7: Area of a Rectangle.

     #Write a function named rectangle_area().

     #Requirements
     #Accept length and width as parameters.
     #Return the area.
     #Validate that both values are positive.
     #Display an error when either value is zero or negative.


     #function to calculate the area of rectangle.

# Function to calculate the area of a rectangle
def rectangle_area(length, width):
    if length <= 0 or width <= 0:
        return "Error: Length and width must be positive numbers."
    return length * width

# Read length and width from the user
length = float(input("Enter the length: "))
width = float(input("Enter the width: "))

# Call the function
result = rectangle_area(length, width)

# Print the result
if isinstance(result, str):
    print(result)
else:
    print("Area of the rectangle:", result)
```

```python
       ##Assignment 8: Area and Circumference of a Circle:

    #Create two functions:

    #circle_area(radius)
    #circle_circumference(radius)

               #Requirements

                 #Accept the radius from the user.
                  #Use math.pi.
                   #Return the area and circumference separately.
                     #Display both values rounded to two decimal places.
import math

# Function to calculate the area of a circle
def circle_area(radius):
    return math.pi * radius ** 2

# Function to calculate the circumference of a circle
def circle_circumference(radius):
    return 2 * math.pi * radius

# Accept the radius from the user
radius = float(input("Enter the radius: "))

# Call the functions
area = circle_area(radius)
circumference = circle_circumference(radius)

# Display the results rounded to two decimal places
print("Area:", round(area, 2))
print("Circumference:", round(circumference, 2))
```

```python
#Assignment 9: Student Grade Calculator
#Write a function named calculate_grade().

#Requirements
  #Accept a student's marks.
  #Return the grade according to the following rules:
  #90–100: A
  #80–89: B
  #70–79: C
  #60–69: D
  #Below 60: F
  #Validate that marks are between 0 and 100.
  # Function to calculate the student's grade.


def calculate_grade(marks):
    if marks < 0 or marks > 100:
        return "Error: Marks must be between 0 and 100."
    elif marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    elif marks >= 60:
        return "D"
    else:
        return "F"

# Accept marks from the user
marks = float(input("Enter the student's marks: "))

# Call the function
grade = calculate_grade(marks)

# Print the result
print("Grade:", grade)
```

```python
#Assignment 10: Simple Interest Calculator
#Write a function named simple_interest().

#Requirements
     #Accept principal, annual interest rate, and time.
    #Return the calculated simple interest.
     #Accept all values from the user.
     #Display the result with two decimal places.
     # Function to calculate simple interest



def simple_interest(principal, rate, time):
    return (principal * rate * time) / 100

# Accept values from the user
principal = float(input("Enter the principal amount: "))
rate = float(input("Enter the annual interest rate (%): "))
time = float(input("Enter the time (in years): "))

# Call the function
interest = simple_interest(principal, rate, time)

# Display the result with two decimal places
print("Simple Interest: {:.2f}".format(interest))
```
