```python
#Assignment 13: Factorial Using a Function
#Write a function named calculate_factorial().

#Requirements
#Accept a non-negative integer.
#Calculate its factorial using a loop.
#Do not use math.factorial().
#Return the factorial.
#Handle negative input.
# Function to calculate factorial
def calculate_factorial(num):
    if num < 0:
        return None

    factorial = 1
    for i in range(1, num + 1):
        factorial *= i

    return factorial

# Accept input from the user
number = int(input("Enter a non-negative integer: "))

# Call the function and display the result
result = calculate_factorial(number)

if result is None:
    print("Factorial is not defined for negative numbers.")
else:
    print("Factorial of", number, "is", result)
```

```python
#Assignment 19: Find Maximum in a List
#Write a function named find_list_maximum().

#Requirements
#Accept a list of numbers.
#Do not use max().
#Return the largest list element.
#Handle an empty list by returning an appropriate message.
# Function to find the maximum element in a list
def find_list_maximum(numbers):
    # Handle an empty list
    if len(numbers) == 0:
        return "The list is empty."

    # Assume the first element is the largest
    maximum = numbers[0]

    # Compare each element with the current maximum
    for num in numbers:
        if num > maximum:
            maximum = num

    return maximum

# Accept input from the user
numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))

# Call the function
result = find_list_maximum(numbers)

# Display the result
print("Result:", result)
```

```python
#Assignment 20: Count Positive, Negative and Zero Values

#Write a function named count_number_types().

           #Requirements
           #Accept a list of integers.
           #Count positive numbers, negative numbers, and zeros.
           #Return all three counts.
           #Use tuple unpacking to receive the returned values.

           # Function to find the maximum element in a list

# Function to count positive, negative, and zero values

def count_number_types(numbers):
    positive = 0
    negative = 0
    zero = 0

    for num in numbers:
        if num > 0:
            positive += 1
        elif num < 0:
            negative += 1
        else:
            zero += 1

    return positive, negative, zero

# Accept input from the user
numbers = list(map(int, input("Enter integers separated by spaces: ").split()))

# Call the function using tuple unpacking
positive_count, negative_count, zero_count = count_number_types(numbers)

# Display the results
print("Positive numbers:", positive_count)
print("Negative numbers:", negative_count)
print("Zero values:", zero_count)
```

```python
#Assignment 32: Function Returning a Dictionary.

#Write a function named analyze_text().

#Requirements
#Accept a string.
#Return a dictionary containing:
#Total characters
#Total words
#Total vowels
#Total digits
#Total spaces
#Display every result clearly.

# Function definition
def analyze_text(text):
    vowels = "aeiouAEIOU"

    result = {
        "Total Characters": len(text),
        "Total Words": len(text.split()),
        "Total Vowels": 0,
        "Total Digits": 0,
        "Total Spaces": 0
    }

    for char in text:
        if char in vowels:
            result["Total Vowels"] += 1
        if char.isdigit():
            result["Total Digits"] += 1
        if char == " ":
            result["Total Spaces"] += 1

    return result

# Accept input from the user
text = input("Enter a string: ")

# Call the function
analysis = analyze_text(text)

# Display the results
print("\nText Analysis")
for key, value in analysis.items():
    print(f"{key}: {value}")
```

```python
#Assignment 33: Function Returning a List
#Write a function named get_even_numbers().

#Requirements
#Accept a starting number and ending number.
#Return a list containing all even numbers in the range.
#Validate that the starting number is smaller than the ending number.


# Function definition
def get_even_numbers(start, end):
    if start >= end:
        return None

    even_numbers = []

    for num in range(start, end + 1):
        if num % 2 == 0:
            even_numbers.append(num)

    return even_numbers

# Accept input from the user
start = int(input("Enter the starting number: "))
end = int(input("Enter the ending number: "))

# Validate and call the function
result = get_even_numbers(start, end)

if result is None:
    print("Error: Starting number must be smaller than the ending number.")
else:
    print("\nEven numbers are:")
    print(result)
```

```python
#Assignment 34: Prime Number Function


#Write a function named is_prime().

#Requirements
#Accept one integer.
#Return True when the number is prime.
#Return False otherwise.
#Use the function to print all prime numbers from 1 to 100.

# Function definition
def is_prime(number):
    if number <= 1:
        return False

    for i in range(2, number):
        if number % i == 0:
            return False

    return True

# Print all prime numbers from 1 to 100
print("Prime numbers from 1 to 100 are:")

for num in range(1, 111):
    if is_prime(num):
        print(num, end=" ")
```

```python
#Assignment 35: Fibonacci Function


#Write a function named generate_fibonacci().

#Requirements
#Accept the number of terms.
#Return the complete Fibonacci series as a list.
#Handle zero and negative term counts.
#Do not use recursion in this assignment.

# Function definition
def generate_fibonacci(terms):
    if terms < 0:
        return "Error: Number of terms cannot be negative."
    elif terms == 0:
        return []

    fibonacci = []

    a, b = 0, 1

    for i in range(terms):
        fibonacci.append(a)
        a, b = b, a + b

    return fibonacci


# Accept input from the user
terms = int(input("Enter the number of terms: "))

# Call the function
result = generate_fibonacci(terms)

# Display the result
print("Fibonacci Series:", result)
```

```python
#Assignment 40: Lambda Maximum of Three Numbers
#Create a lambda expression that returns the largest of three numbers.

#Requirements
#Do not use max().
#Use nested conditional expressions.
#Accept all values from the user.
# Lambda expression to find the largest of three numbers
maximum = lambda a, b, c: a if a >= b and a >= c else (b if b >= c else c)

# Accept input from the user
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
num3 = float(input("Enter the third number: "))

# Call the lambda function
result = maximum(num1, num2, num3)

# Display the result
print("Largest number:", result)
```

```python
#Assignment 41: Sort Tuples Using Lambda


#Given the following list:

#students = [
   # ("Ravi", 82),
   # ("Anitha", 95),
   # ("Kiran", 76),
   # ("Meena", 89)]
#Requirements
#Sort students by marks in ascending order.
#Sort students by marks in descending order.
#Use sorted() with a lambda as the key.
#Print both results.

# List of students
students = [
    ("Ravi", 82),
    ("Anitha", 95),
    ("Kiran", 76),
    ("Meena", 89)
]

# Sort by marks in ascending order
ascending = sorted(students, key=lambda student: student[1])

# Sort by marks in descending order
descending = sorted(students, key=lambda student: student[1], reverse=True)

# Display the results
print("Students sorted by marks (Ascending):")
for student in ascending:
    print(student)

print("\nStudents sorted by marks (Descending):")
for student in descending:
    print(student)
```

```python
#Assignment 42: Sort Dictionaries by Age


#Given a list of dictionaries containing names and ages, sort the list by age.

#Requirements


#Use a lambda expression as the sorting key.
#Display the youngest student first.
#Display the oldest student first in another result.
#Sample Data
#students = [
    #{"name": "Ravi", "age": 24},
    #{"name": "Anu", "age": 21},
    #{"name": "Kiran", "age": 27}]


# List of dictionaries
students = [
    {"name": "Ravi", "age": 24},
    {"name": "Anu", "age": 21},
    {"name": "Kiran", "age": 27}
]

# Sort by age (youngest to oldest)
youngest_first = sorted(students, key=lambda student: student["age"])

# Sort by age (oldest to youngest)
oldest_first = sorted(students, key=lambda student: student["age"], reverse=True)

# Display the results
print("Youngest Student First:")
for student in youngest_first:
    print(student)

print("\nOldest Student First:")
for student in oldest_first:
    print(student)
```

```python
#Assignment 43: Sort Words by Length

#Write a program that accepts a sentence.

#Requirements
#Split the sentence into words.
#Sort words according to their length.
#Use a lambda as the key.
#When two words have the same length, sort them alphabetically.


#Hint


#Use a tuple as the sorting key:

#key=lambda word: (len(word), word)

# Accept a sentence from the user
sentence = input("Enter a sentence: ")

# Split the sentence into words
words = sentence.split()

# Sort words by length, then alphabetically
sorted_words = sorted(words, key=lambda word: (len(word), word))

# Display the result
print("\nWords sorted by length:")
for word in sorted_words:
    print(word)
```

```python
#Assignment 45: Lambda Discount Calculator
#Create a lambda expression that accepts product price and discount percentage.

#Requirements
#Calculate the final price after discount.
#Validate inputs before calling the lambda.
#Display the discount amount and final price.

# Lambda expression to calculate the final price
calculate_final_price = lambda price, discount: price - (price * discount / 100)

# Accept input from the user
price = float(input("Enter product price: "))
discount = float(input("Enter discount percentage: "))

# Validate inputs
if price < 0:
    print("Error: Product price cannot be negative.")
elif discount < 0 or discount > 100:
    print("Error: Discount percentage must be between 0 and 100.")
else:
    # Calculate discount amount
    discount_amount = price * discount / 100

    # Call the lambda function
    final_price = calculate_final_price(price, discount)

    # Display the results
    print("\nDiscount Amount:", discount_amount)
    print("Final Price:", final_price)
```

```python
#Assignment 46: Square Numbers Using map()
#Given a list of numbers, generate their squares using map().

#Requirements
#Accept numbers from the user.
#Use map() with a lambda expression.
#Convert the map object into a list.
#Print original and squared lists.
#Do not use a loop or list comprehension for transformation.

# Accept numbers from the user
numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))

# Use map() with a lambda expression to calculate squares
squared_numbers = list(map(lambda x: x ** 2, numbers))

# Display the results
print("Original List:", numbers)
print("Squared List:", squared_numbers)
```

```python
#Assignment 48: Convert Celsius List to Fahrenheit


#Given a list of Celsius temperatures, convert every value to Fahrenheit.

#Requirements
#Use map() and a lambda expression.
#Round every converted value to two decimal places.
#Print the Celsius and Fahrenheit lists.
# Accept Celsius temperatures from the user
celsius = list(map(float, input("Enter Celsius temperatures separated by spaces: ").split()))

# Convert Celsius to Fahrenheit using map() and lambda
fahrenheit = list(map(lambda c: round((c * 9/5) + 32, 2), celsius))

# Display the results
print("Celsius List:", celsius)
print("Fahrenheit List:", fahrenheit)
```

```python
#Assignment 49: Convert Strings to Uppercase

#Accept a list of names from the user.

#Requirements
#Convert every name to uppercase using map().
#Remove extra spaces from each name.
#Print the cleaned list.
#Hint
#Use string methods inside a lambda.

# Accept names from the user
names = input("Enter names separated by commas: ").split(",")

# Remove extra spaces and convert to uppercase
cleaned_names = list(map(lambda name: name.strip().upper(), names))

# Display the results
print("Original List:", names)
print("Cleaned List:", cleaned_names)
```

```python
#Assignment 50: Find Length of Every Word


#Accept a sentence from the user.

#Requirements
#Split the sentence into words.
#Use map() to calculate the length of every word.
#Combine every word and its length into tuples.


# Accept a sentence from the user
sentence = input("Enter a sentence: ")

# Split the sentence into words
words = sentence.split()

# Use map() to calculate the length of each word
lengths = list(map(len, words))

# Combine each word with its length into tuples
result = list(zip(words, lengths))

# Display the result
print("Word Lengths:", result)
```
