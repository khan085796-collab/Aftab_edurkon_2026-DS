```python
#Assignment 13: Factorial Using a Function
#Write a function named calculate_factorial().

#Requirements
#Accept a non-negative integer.
#Calculate its factorial using a loop.
#Do not use math.factorial().
#Return the factorial.
#Handle negative input.
# Function to calculate factorial
def calculate_factorial(num):
    if num < 0:
        return None

    factorial = 1
    for i in range(1, num + 1):
        factorial *= i

    return factorial

# Accept input from the user
number = int(input("Enter a non-negative integer: "))

# Call the function and display the result
result = calculate_factorial(number)

if result is None:
    print("Factorial is not defined for negative numbers.")
else:
    print("Factorial of", number, "is", result)
```

```python
#Assignment 19: Find Maximum in a List
#Write a function named find_list_maximum().

#Requirements
#Accept a list of numbers.
#Do not use max().
#Return the largest list element.
#Handle an empty list by returning an appropriate message.
# Function to find the maximum element in a list
def find_list_maximum(numbers):
    # Handle an empty list
    if len(numbers) == 0:
        return "The list is empty."

    # Assume the first element is the largest
    maximum = numbers[0]

    # Compare each element with the current maximum
    for num in numbers:
        if num > maximum:
            maximum = num

    return maximum

# Accept input from the user
numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))

# Call the function
result = find_list_maximum(numbers)

# Display the result
print("Result:", result)
```

```python
#Assignment 32: Function Returning a Dictionary.

#Write a function named analyze_text().

#Requirements
#Accept a string.
#Return a dictionary containing:
#Total characters
#Total words
#Total vowels
#Total digits
#Total spaces
#Display every result clearly.

# Function definition
def analyze_text(text):
    vowels = "aeiouAEIOU"

    result = {
        "Total Characters": len(text),
        "Total Words": len(text.split()),
        "Total Vowels": 0,
        "Total Digits": 0,
        "Total Spaces": 0
    }

    for char in text:
        if char in vowels:
            result["Total Vowels"] += 1
        if char.isdigit():
            result["Total Digits"] += 1
        if char == " ":
            result["Total Spaces"] += 1

    return result

# Accept input from the user
text = input("Enter a string: ")

# Call the function
analysis = analyze_text(text)

# Display the results
print("\nText Analysis")
for key, value in analysis.items():
    print(f"{key}: {value}")
```

```python
#Assignment 33: Function Returning a List
#Write a function named get_even_numbers().

#Requirements
#Accept a starting number and ending number.
#Return a list containing all even numbers in the range.
#Validate that the starting number is smaller than the ending number.


# Function definition
def get_even_numbers(start, end):
    if start >= end:
        return None

    even_numbers = []

    for num in range(start, end + 1):
        if num % 2 == 0:
            even_numbers.append(num)

    return even_numbers

# Accept input from the user
start = int(input("Enter the starting number: "))
end = int(input("Enter the ending number: "))

# Validate and call the function
result = get_even_numbers(start, end)

if result is None:
    print("Error: Starting number must be smaller than the ending number.")
else:
    print("\nEven numbers are:")
    print(result)
```

```python
#Assignment 34: Prime Number Function


#Write a function named is_prime().

#Requirements
#Accept one integer.
#Return True when the number is prime.
#Return False otherwise.
#Use the function to print all prime numbers from 1 to 100.

# Function definition
def is_prime(number):
    if number <= 1:
        return False

    for i in range(2, number):
        if number % i == 0:
            return False

    return True

# Print all prime numbers from 1 to 100
print("Prime numbers from 1 to 100 are:")

for num in range(1, 111):
    if is_prime(num):
        print(num, end=" ")
```

```python
#Assignment 35: Fibonacci Function


#Write a function named generate_fibonacci().

#Requirements
#Accept the number of terms.
#Return the complete Fibonacci series as a list.
#Handle zero and negative term counts.
#Do not use recursion in this assignment.

# Function definition
def generate_fibonacci(terms):
    if terms < 0:
        return "Error: Number of terms cannot be negative."
    elif terms == 0:
        return []

    fibonacci = []

    a, b = 0, 1

    for i in range(terms):
        fibonacci.append(a)
        a, b = b, a + b

    return fibonacci


# Accept input from the user
terms = int(input("Enter the number of terms: "))

# Call the function
result = generate_fibonacci(terms)

# Display the result
print("Fibonacci Series:", result)
```

```python
#Assignment 68: Filter Palindrome Words

#Accept a sentence from the user.

#Requirements
#Split it into words.
#Use filter() to find palindrome words.
#Ignore case and punctuation.
#Write a separate function for palindrome checking.



import string

# Function to check whether a word is a palindrome
def is_palindrome(word):
    # Remove punctuation and convert to lowercase
    cleaned_word = word.strip(string.punctuation).lower()

    # Check if the cleaned word is a palindrome
    return cleaned_word != "" and cleaned_word == cleaned_word[::-1]

# Accept a sentence from the user
sentence = input("Enter a sentence: ")

# Split the sentence into words
words = sentence.split()

# Filter palindrome words
palindrome_words = list(filter(is_palindrome, words))

# Display the results
print("Original Words :", words)
print("Palindrome Words:", palindrome_words)
```

```python
#Assignment 69: Filter Valid Email Addresses


#Given a list of email strings:

#Requirements
#Use filter().
#A basic valid email should:
#Contain exactly one @.
#Have characters before @.
#Contain a dot after @.
#Return only emails satisfying these conditions.



# Function to check whether an email is valid
def is_valid_email(email):
    # Must contain exactly one '@'
    if email.count("@") != 1:
        return False

    username, domain = email.split("@")

    # Username must not be empty
    if username == "":
        return False

    # Domain must contain at least one dot
    if "." not in domain:
        return False

    return True

# List of email addresses
emails = [
    "ravi@gmail.com",
    "meena@yahoo",
    "@outlook.com",
    "kiran@hotmail.com",
    "anil@@gmail.com",
    "john.doe@example.org",
    "abc@domain",
    "sita@company.co.in"
]

# Filter valid email addresses
valid_emails = list(filter(is_valid_email, emails))

# Display the results
print("Original Emails:")
print(emails)

print("\nValid Emails:")
print(valid_emails)
```

```python
#Assignment 81: Calculate Factorial Using reduce()



#Write a function that calculates factorial using reduce().

#Requirements
#Accept a non-negative integer.
#Generate numbers from 1 to the given number.
#Multiply them using reduce().
#Return 1 when the input is 0.
#Reject negative values.


from functools import reduce

# Function to calculate factorial
def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers."
    elif n == 0:
        return 1
    else:
        return reduce(lambda x, y: x * y, range(1, n + 1))

# Accept input from the user
number = int(input("Enter a non-negative integer: "))

# Display the result
result = factorial(number)
print("Factorial:", result)
```

```python
#Assignment 92: Closure-Based Bank Balance


#Write a function named create_bank_account().

#Requirements
##Store the balance in the outer function.
#Define inner functions for:
#Deposit
#Withdrawal
#Balance enquiry
#Use nonlocal to update the balance.
#Reject withdrawals greater than the available balance.
#Return all three inner functions.

# Assignment 92: Closure-Based Bank Balance

# Outer function
def create_bank_account(initial_balance):

    balance = initial_balance

    # Deposit function
    def deposit(amount):
        nonlocal balance
        if amount > 0:
            balance += amount
            print(f"Deposited: ₹{amount:.2f}")
        else:
            print("Invalid deposit amount.")

    # Withdrawal function
    def withdraw(amount):
        nonlocal balance
        if amount <= 0:
            print("Invalid withdrawal amount.")
        elif amount > balance:
            print("Insufficient balance.")
        else:
            balance -= amount
            print(f"Withdrawn: ₹{amount:.2f}")

    # Balance enquiry function
    def check_balance():
        print(f"Current Balance: ₹{balance:.2f}")

    # Return all three inner functions
    return deposit, withdraw, check_balance


# Create a bank account with an initial balance
deposit, withdraw, check_balance = create_bank_account(5000)

# Perform operations
check_balance()

deposit(1500)
check_balance()

withdraw(2000)
check_balance()

withdraw(6000)   # Exceeds available balance
check_balance()
```

```python
#Assignment 93: Function Decorator


#Create a decorator named execution_logger().

#Requirements
#The decorator should print a message before the original function executes.
#It should print another message after execution.
#Apply it to an addition function.
#Preserve the original function's return value.




# Decorator function
def execution_logger(func):

    def wrapper(*args, **kwargs):
        print("Starting function execution...")

        # Call the original function
        result = func(*args, **kwargs)

        print("Function execution completed.")

        # Return the original function's result
        return result

    return wrapper


# Apply the decorator
@execution_logger
def add(a, b):
    return a + b


# Accept input from the user
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

# Call the decorated function
sum_result = add(num1, num2)

# Display the result
print("Sum:", sum_result)
```

```python
#Part 6: Total Salary Expense
#Use reduce() to calculate:

#Current total salary
#Revised total salary
#Total additional salary expense

from functools import reduce

# Sample Employee Data
employees = [
    {
        "id": 101,
        "name": "Ravi",
        "department": "IT",
        "salary": 60000,
        "performance_score": 88,
        "experience": 5
    },
    {
        "id": 102,
        "name": "Priya",
        "department": "HR",
        "salary": 55000,
        "performance_score": 92,
        "experience": 4
    },
    {
        "id": 103,
        "name": "Amit",
        "department": "Finance",
        "salary": 70000,
        "performance_score": 85,
        "experience": 6
    }
]

# Calculate revised salaries using map()
revised_salaries = list(map(
    lambda employee: {
        "name": employee["name"],
        "old_salary": employee["salary"],
        "new_salary":
            employee["salary"] * 1.15 if employee["performance_score"] >= 90
            else employee["salary"] * 1.10 if employee["performance_score"] >= 80
            else employee["salary"] * 1.05 if employee["performance_score"] >= 70
            else employee["salary"]
    },
    employees
))

# Current total salary
current_total = reduce(
    lambda total, employee: total + employee["salary"],
    employees,
    0
)

# Revised total salary
revised_total = reduce(
    lambda total, employee: total + employee["new_salary"],
    revised_salaries,
    0
)

# Additional salary expense
additional_expense = revised_total - current_total

# Display results
print("Current Total Salary :", current_total)
print("Revised Total Salary :", revised_total)
print("Additional Expense   :", additional_expense)
```

```python
#Part 9: Closure-Based Bonus Calculator
#Create a function named create_bonus_calculator().

#The outer function should accept a bonus percentage.
#The inner function should accept a salary.
#Return the calculated bonus.
#Create separate bonus calculators for 5%, 10%, and 15%.

# Function to create a bonus calculator
def create_bonus_calculator(bonus_percentage):
    def calculate_bonus(salary):
        return salary * (bonus_percentage / 100)
    return calculate_bonus

# Create bonus calculators
bonus_5 = create_bonus_calculator(5)
bonus_10 = create_bonus_calculator(10)
bonus_15 = create_bonus_calculator(15)

# Sample salary
salary = 60000

# Display bonuses
print("Salary:", salary)
print("5% Bonus :", bonus_5(salary))
print("10% Bonus:", bonus_10(salary))
print("15% Bonus:", bonus_15(salary))
```

```python
#Part 10: Nested Report Generator
#Create an outer function named generate_department_report().

#Inside it, define nested functions to calculate:

#Number of employees
#Total salary
#Average salary
#Average performance score
#Return the complete department report as a dictionary.

# Sample Employee Data
employees = [
    {
        "id": 101,
        "name": "Ravi",
        "department": "IT",
        "salary": 60000,
        "performance_score": 88,
        "experience": 5
    },
    {
        "id": 102,
        "name": "Priya",
        "department": "HR",
        "salary": 55000,
        "performance_score": 92,
        "experience": 4
    },
    {
        "id": 103,
        "name": "Amit",
        "department": "Finance",
        "salary": 70000,
        "performance_score": 85,
        "experience": 6
    }
]

# Outer function
def generate_department_report(department):
    # Filter employees by department
    dept_employees = list(
        filter(lambda emp: emp["department"].lower() == department.lower(), employees)
    )

    # Nested function: Number of employees
    def employee_count():
        return len(dept_employees)

    # Nested function: Total salary
    def total_salary():
        return sum(emp["salary"] for emp in dept_employees)

    # Nested function: Average salary
    def average_salary():
        if employee_count() == 0:
            return 0
        return total_salary() / employee_count()

    # Nested function: Average performance score
    def average_performance():
        if employee_count() == 0:
            return 0
        total_score = sum(emp["performance_score"] for emp in dept_employees)
        return total_score / employee_count()

    # Return report as a dictionary
    return {
        "Department": department,
        "Number of Employees": employee_count(),
        "Total Salary": total_salary(),
        "Average Salary": average_salary(),
        "Average Performance Score": average_performance()
    }

# Example
report = generate_department_report("IT")

print("Department Report")
for key, value in report.items():
    print(f"{key}: {value}")
```

```python
#Part 12: Menu-Driven Program
#Create a menu containing the following options:

#1. Display all employees
#2. Display employee names
#3. Display high performers
#4. Search by department
#5. Calculate revised salaries
#6. Calculate total salary expense
#7. Find highest-paid employee
#8. Sort employees
#9. Generate department report
#10. Exit

from functools import reduce

# Sample Employee Data
employees = [
    {
        "id": 101,
        "name": "Ravi",
        "department": "IT",
        "salary": 60000,
        "performance_score": 88,
        "experience": 5
    },
    {
        "id": 102,
        "name": "Priya",
        "department": "HR",
        "salary": 55000,
        "performance_score": 92,
        "experience": 4
    },
    {
        "id": 103,
        "name": "Amit",
        "department": "Finance",
        "salary": 70000,
        "performance_score": 85,
        "experience": 6
    }
]

# 1. Display Employees
def display_employees():
    for emp in employees:
        print("-" * 40)
        for key, value in emp.items():
            print(f"{key.capitalize()}: {value}")
    print("-" * 40)

# 2. Display Employee Names
def display_employee_names():
    names = list(map(lambda emp: emp["name"], employees))
    print(names)

# 3. Display High Performers
def display_high_performers():
    high = list(filter(lambda emp: emp["performance_score"] >= 80, employees))
    for emp in high:
        print(emp)

# 4. Search by Department
def filter_by_department(department):
    result = list(filter(
        lambda emp: emp["department"].lower() == department.lower(),
        employees
    ))

    if result:
        for emp in result:
            print(emp)
    else:
        print("No employees found.")

# 5. Calculate Revised Salaries
def revised_salaries():
    revised = list(map(
        lambda emp: {
            "name": emp["name"],
            "old_salary": emp["salary"],
            "new_salary":
                emp["salary"] * 1.15 if emp["performance_score"] >= 90
                else emp["salary"] * 1.10 if emp["performance_score"] >= 80
                else emp["salary"] * 1.05 if emp["performance_score"] >= 70
                else emp["salary"]
        },
        employees
    ))

    for emp in revised:
        print(emp)

    return revised

# 6. Total Salary Expense
def total_salary_expense():
    revised = revised_salaries()

    current_total = reduce(
        lambda total, emp: total + emp["salary"],
        employees,
        0
    )

    revised_total = reduce(
        lambda total, emp: total + emp["new_salary"],
        revised,
        0
    )

    print("\nCurrent Total Salary :", current_total)
    print("Revised Total Salary :", revised_total)
    print("Additional Expense   :", revised_total - current_total)

# 7. Highest Paid Employee
def highest_paid_employee():
    highest = reduce(
        lambda e1, e2: e1 if e1["salary"] > e2["salary"] else e2,
        employees
    )

    print(highest)

# 8. Sort Employees
def sort_employees():
    print("\n1. Name")
    print("2. Salary")
    print("3. Performance Score")
    print("4. Experience")

    choice = input("Choose sorting option: ")

    if choice == "1":
        result = sorted(employees, key=lambda emp: emp["name"])
    elif choice == "2":
        result = sorted(employees, key=lambda emp: emp["salary"])
    elif choice == "3":
        result = sorted(employees, key=lambda emp: emp["performance_score"])
    elif choice == "4":
        result = sorted(employees, key=lambda emp: emp["experience"])
    else:
        print("Invalid choice.")
        return

    for emp in result:
        print(emp)

# 9. Department Report
def generate_department_report(department):
    dept = list(filter(
        lambda emp: emp["department"].lower() == department.lower(),
        employees
    ))

    if not dept:
        print("No employees found.")
        return

    report = {
        "Department": department,
        "Number of Employees": len(dept),
        "Total Salary": sum(emp["salary"] for emp in dept),
        "Average Salary": sum(emp["salary"] for emp in dept) / len(dept),
        "Average Performance":
            sum(emp["performance_score"] for emp in dept) / len(dept)
    }

    for key, value in report.items():
        print(f"{key}: {value}")

# 10. Menu
while True:
    print("\n===== Employee Performance Analysis System =====")
    print("1. Display all employees")
    print("2. Display employee names")
    print("3. Display high performers")
    print("4. Search by department")
    print("5. Calculate revised salaries")
    print("6. Calculate total salary expense")
    print("7. Find highest-paid employee")
    print("8. Sort employees")
    print("9. Generate department report")
    print("10. Exit")

    option = input("Enter your choice: ")

    if option == "1":
        display_employees()

    elif option == "2":
        display_employee_names()

    elif option == "3":
        display_high_performers()

    elif option == "4":
        department = input("Enter department: ")
        filter_by_department(department)

    elif option == "5":
        revised_salaries()

    elif option == "6":
        total_salary_expense()

    elif option == "7":
        highest_paid_employee()

    elif option == "8":
        sort_employees()

    elif option == "9":
        department = input("Enter department: ")
        generate_department_report(department)

    elif option == "10":
        print("Thank you for using the Employee Performance Analysis System.")
        break

    else:
        print("Invalid choice! Please try again.")
```
