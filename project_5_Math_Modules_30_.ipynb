```python
##Calculate the sine, cosine, and tangent of 45 degrees.


# Convert 45 degrees to radians
angle = math.radians(45)

# Calculate sine, cosine, and tangent
sin_value = math.sin(angle)
cos_value = math.cos(angle)
tan_value = math.tan(angle)

print("Sin(45°) =", sin_value)
print("Cos(45°) =", cos_value)
print("Tan(45°) =", tan_value)
```

```python
### Calculate compound interest using math.pow().
principle = 10000
rate = 2
time = 2
compound_interest = principle * (math.pow((1 + rate / 100), time)) - principle
print(compound_interest)
```

```python
### Find the area and circumference of a circle using math.pi.
radius = 5
area = math.pi * radius ** 2
circumference = 2 * math.pi * radius
print(area)
print(circumference)
```
